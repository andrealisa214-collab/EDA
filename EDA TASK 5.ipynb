```python
def catergorize_urgency(admission):
    if admission=="Emercency":
       return "immidate care"
    elif admission=="urgent":
        return "high priority"
    else:
        return "planned care"

df["Priority_level"]=df["Admission Type"].apply(catergorize_urgency)
```

```python
df
```

```python
df["Hospital stay days"]=(df["Discharge Date"]-df["Date of Admission"]).dt.days
```
