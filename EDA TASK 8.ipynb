```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load dataset
df=pd.read_csv("StudentsPerformance.csv")
```

```python
df
```

```python
print(df.head())#
```

```python
print(df.columns)#
```

```python
perental_avg=df.groupby("parental level of education")[
    ["math score", "reading score",	"writing score"]
    ].mean()
print("/Average scores by perental education:")
print("perental_avg")
perental_avg.plot (
    kind="bar",
   figsize=(12,6)
 ) 

plt.title("Average Test Scores by Parental Education Level")
plt.xlabel("Parental Education Level")
plt.ylabel("Average Score")
plt.xticks(rotation=30)
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```

```python
perental_avg=df.groupby("lunch")[
    ["math score", "reading score",	"writing score"]
    ].mean()
print("/Average scores by lunch:")
print("lunch")
perental_avg.plot (
    kind="bar",
   figsize=(12,6)
 ) 

plt.title("Average Test Scores by lunch")
plt.xlabel("lunch")
plt.ylabel("Average Score")
plt.xticks(rotation=30)
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(
    data=df,
    x="test preparation course",
    y="reading score"
)

plt.title("Reading Score by Test Preparation")
plt.xlabel("Test Preparation Course")
plt.ylabel("Reading Score")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="math score",
    y="reading score"
)

plt.title("math score vs Reading Score")
plt.xlabel("math score")
plt.ylabel("Reading Score")
plt.grid(True)
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="math score",
    y="writing score"
)

plt.title("math score vs writing score")
plt.xlabel("math score")
plt.ylabel("writing score")
plt.grid(True)
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="writing score",
    y="reading score"
)

plt.title("writing score vs reading score")
plt.xlabel("writing score")
plt.ylabel("reading score")
plt.grid(True)
plt.show()
```
